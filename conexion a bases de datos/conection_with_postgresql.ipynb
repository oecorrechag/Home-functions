```python
import os
import pandas as pd
import re
from sqlalchemy import create_engine
from dotenv import load_dotenv

load_dotenv()

def sync_insert_dataframe(df, table_name, if_exists='replace'):
    # Crear engine síncrono
    database_url = os.getenv('DATABASE_URL')
    engine = create_engine(database_url)
    
    # Usar el método to_sql de pandas
    df.to_sql(
        name=table_name,
        con=engine,
        if_exists=if_exists,  # 'fail', 'replace', 'append'
        index=False,
        method='multi'
    )
    
    engine.dispose()
    print(f"DataFrame insertado en {table_name}")

# Ejemplo de uso
df = pd.DataFrame({
    'producto': ['Laptop', 'Mouse', 'Teclado'],
    'precio': [999.99, 25.50, 75.00],
    'stock': [10, 50, 30]
})

sync_insert_dataframe(df, 'productos', if_exists='replace')
```
